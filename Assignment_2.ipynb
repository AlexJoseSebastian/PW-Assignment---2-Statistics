Question 1: What is hypothesis testing in statistics?
- Answer: Hypothesis testing is a statistical method used to make a decision using sample data. It helps us decide whether an assumption about a population is true or false.

Question 2: What is the null hypothesis, and how does it differ from the alternative hypothesis?
- Answer:
- Null hypothesis (H₀): There is no effect or no difference.
- Alternative hypothesis (H₁): There is an effect or a difference.

- Example
- H₀: Mean = 50
- H₁: Mean ≠ 50

Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.
- Answer: The significance level (α) is the probability of rejecting a true null hypothesis. Common value: 0.05 (5%)

- If p-value ≤ α, reject H₀

- If p-value > α, fail to reject H₀

Question 4: What are Type I and Type II errors? Give examples of each.
- Answer:
- Type I Error: Rejecting a true null hypothesis.

- Example: Saying a medicine works when it actually does not.

- Type II Error: Not rejecting a false null hypothesis.

- Example: Saying a medicine does not work when it actually works.

Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.
- Answer:
- Z-test:
  - Population std deviation known.
  - Large sample(n>=30).
  - Uses normal distribution.

- T-test:
  - Std deviation unknown
  - Small sample(n < 30)
  - Uses t-distribution.

Question 6: Write a Python program to generate a binomial distribution with n=10 and p=0.5, then plot its histogram.

```python
import random
import matplotlib.pyplot as plt

data = [random.binomialvariate(10,0.5)  for _ in range(1000) ]

plt.hist(data)
plt.title('Binomial Distribution of (n=10,p=0.5)')
plt.xlabel('Numer of success')
plt.ylabel('Frequency')
plt.show()
```

Question 7: Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results.
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9].

```python
import numpy as np
from scipy import stats

sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

mean = np.mean(sample_data)
std = np.std(sample_data, ddof=1)
n = len(sample_data)

z = (mean - 50) / (std / np.sqrt(n))
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print("Z-score:", z)
print("P-value:", p_value)
```

Question 8: Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

data = np.random.normal(50, 5, 1000)

mean = np.mean(data)
std = np.std(data, ddof=1)

ci = stats.norm.interval(0.95, mean, std/np.sqrt(len(data)))

print("95% Confidence Interval:", ci)

plt.hist(data)
plt.title("Normal Distribution")
plt.show()
```

Question 9: Write a Python function to calculate the Z-scores from a dataset and
visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.


```python
import numpy as np
import matplotlib.pyplot as plt

data = [10, 20, 30, 40, 50]

mean = np.mean(data)
std = np.std(data)

z_scores = [(x - mean) / std for x in data]

plt.hist(z_scores)
plt.title("Z-score Histogram")
plt.show()

print(z_scores)
```
